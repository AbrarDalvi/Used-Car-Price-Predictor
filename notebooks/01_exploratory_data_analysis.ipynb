# Used Car Resale Price Prediction - Exploratory Data Analysis (EDA)
**Objective**: Understand the distribution of car prices, identify relationships between features, and uncover data quality issues (missing values, duplicates, mixed-unit columns).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visual settings
sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', None)

# Load dataset
df = pd.read_csv('../data/Car details v3.csv')
print(f'Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns')
df.head()
```

### 1. Data Schema & Missing Values

```python
df.info()
print('\n--- Missing Values Count ---')
print(df.isnull().sum()[df.isnull().sum() > 0])
print(f'\n--- Duplicate Rows: {df.duplicated().sum()} ---')
```

### 2. Distribution of Target Variable: `selling_price`

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df['selling_price'] / 1e5, kde=True, ax=ax[0], color='royalblue', bins=35)
ax[0].set_title('Selling Price Distribution (Lakh Rs)', fontsize=12, fontweight='bold')
ax[0].set_xlabel('Selling Price (Lakh Rs)')

sns.boxplot(x=df['selling_price'] / 1e5, ax=ax[1], color='lightsteelblue')
ax[1].set_title('Selling Price Boxplot (Outliers Check)', fontsize=12, fontweight='bold')
ax[1].set_xlabel('Selling Price (Lakh Rs)')
plt.show()
```

### 3. Categorical Variables vs Selling Price

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.boxplot(data=df, x='fuel', y=df['selling_price']/1e5, ax=axes[0], hue='fuel', legend=False, palette='Set2')
axes[0].set_title('Price vs Fuel')
axes[0].set_ylim(0, 35)

sns.boxplot(data=df, x='seller_type', y=df['selling_price']/1e5, ax=axes[1], hue='seller_type', legend=False, palette='Set3')
axes[1].set_title('Price vs Seller Type')
axes[1].set_ylim(0, 35)

sns.boxplot(data=df, x='transmission', y=df['selling_price']/1e5, ax=axes[2], hue='transmission', legend=False, palette='Pastel1')
axes[2].set_title('Price vs Transmission')
axes[2].set_ylim(0, 45)
plt.show()
```
